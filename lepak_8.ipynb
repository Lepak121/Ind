```python
# 1. Встановлення необхідних бібліотек
!pip install mlxtend -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# ==========================================
# 1. ЗАВАНТАЖЕННЯ ТА ПОПЕРЕДНЯ ОБРОБКА ДАНИХ
# ==========================================
# Переконайтеся, що файл Basket_data.csv завантажено у панель файлів Google Colab
try:
    with open('Basket_data.csv', 'r') as f:
        lines = f.readlines()
    transactions = [line.strip().split(',') for line in lines]
except FileNotFoundError:
    print("Файл Basket_data.csv не знайдено. Використовуються тестові дані.")
    transactions = [
        ['mineral water', 'whole wheat pasta', 'olive oil'],
        ['fromage blanc', 'honey'],
        ['spaghetti', 'tomato sauce', 'ground beef'],
        ['light cream', 'chicken'],
        ['pasta', 'escalope']
    ] * 1500

# Очищення від порожніх значень
transactions = [[item.strip() for item in t if item and item.strip()] for t in transactions]
print(f"Кількість оброблених транзакцій (чекiв): {len(transactions)}")

# ==========================================
# 2. ЗАСТОСУВАННЯ АЛГОРИТМУ APRIORI
# ==========================================
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_trans = pd.DataFrame(te_ary, columns=te.columns_)

frequent_itemsets = apriori(df_trans, min_support=0.003, use_colnames=True, max_len=3)
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=3.0)
rules = rules[rules['confidence'] >= 0.2]
rules = rules.sort_values(by='lift', ascending=False).reset_index(drop=True)

print(f"Сформовано валідних асоціативних правил: {len(rules)}\n")

# ==========================================
# 3. ВІЗУАЛІЗАЦІЯ СТРУКТУРИ ПРАВИЛ
# ==========================================

# А. Орієнтований граф асоціацій (Top-20 за показником Lift)
top_20_rules = rules.head(20)

plt.figure(figsize=(10, 8), dpi=100)
G = nx.DiGraph()

for idx, row in top_20_rules.iterrows():
    ant = ', '.join(list(row['antecedents']))
    con = ', '.join(list(row['consequents']))
    G.add_edge(ant, con, weight=row['lift'])

pos = nx.spring_layout(G, k=1.2, seed=42)
nx.draw(
    G, pos, with_labels=True,
    node_color='#aec7e8', node_size=2000,
    edge_color='#ff7f0e', linewidths=1.5,
    font_size=8, font_weight='bold',
    arrows=True, arrowsize=15
)
plt.title('Орієнтований граф асоціацій (Top-20 за показником Lift)', fontsize=11)
plt.tight_layout()
plt.show()

# Б. Кругова діаграма часток підтримки (Support) для топ-10 правил
top_10_rules = rules.head(10).copy()
rule_labels = [f"Правило {i+1}" for i in range(len(top_10_rules))]

plt.figure(figsize=(8, 6), dpi=100)
plt.pie(
    top_10_rules['support'],
    labels=rule_labels,
    autopct='%1.1f%%',
    startangle=140,
    colors=plt.cm.Paired.colors
)
plt.title('Відносний розподіл підтримки (Support) топ-10 асоціативних правил', fontsize=11)
plt.tight_layout()
plt.show()

# В. Порівняння рівнів Support та Lift для топ-10 правил (як на вашому скріншоті)
top_10_rules['rule_str'] = top_10_rules.apply(
    lambda row: f"{', '.join(list(row['antecedents']))} -> {', '.join(list(row['consequents']))}",
    axis=1
)

fig, axes = plt.subplots(1, 2, figsize=(14, 6), dpi=100, sharey=True)

# Ліва панель: Рівень підтримки (%)
axes[0].barh(top_10_rules['rule_str'], top_10_rules['support'] * 100, color='#1f77b4')
axes[0].set_title('Рівень підтримки (Support)', fontsize=11)
axes[0].set_xlabel('Підтримка, Support (%)', fontsize=9)
axes[0].invert_yaxis()  # Найкраще правило зверху

# Права панель: Сила зв'язку (Lift)
axes[1].barh(top_10_rules['rule_str'], top_10_rules['lift'], color='#ff7f0e')
axes[1].set_title("Сила асоціативного зв'язку (Lift)", fontsize=11)
axes[1].set_xlabel('Підйом, Lift', fontsize=9)

for ax in axes:
    ax.grid(True, linestyle=':', alpha=0.6)

plt.suptitle('Порівняння рівнів Support та Lift для топ-10 асоціативних правил', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```
